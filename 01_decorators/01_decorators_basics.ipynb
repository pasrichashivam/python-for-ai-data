## Python Decorators
<b>Decorators let you add extra behavior to a function, without changing the function's code.<br>
A decorator is a function that takes another function as input and returns a new function.

</b>

```python
from datetime import datetime
import time
from functools import wraps
```

### Simple decorator
- Demo decorator to show the flow of call stack.

```python
def my_decorator(func):
    def wrapper():
        print(f"Funtionality before the '{func.__name__}' function is called.")
        func()  # calling greet()
        print(f"Funtionality after the '{func.__name__}' function is called.")

    return wrapper

@my_decorator
def greet():
    print('Hello Shivam!')

greet()
```

### Time_it decorator that logs the time to execute a function.

```python
def time_it(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        start = time.time()
        func(*args, **kwargs)
        print(f'{func.__name__} took {str(round(time.time() - start, 3))} seconds')

    return wrapper

@time_it
def stress_method(seconds):
    time.sleep(seconds)
```

```python
stress_method(3.4)
```

### Decorator to Add Guardrail that executes function only 7 AM to 10 PM

```python
# To call the function only between the 7 AM to 10 PM
def call_between_7_to_22(func):
    def wrapper(*args, **kwargs):
        if 7 <= datetime.now().hour < 22:
            func(*args, **kwargs)
        else:
            print(f'SKIP: Function {func.__name__} can be called between 7 AM to 10 PM')

    return wrapper

@call_between_7_to_22
def bank_transaction():
    print("Transaction Done!")
```

```python
bank_transaction()
```

### Decorator that takes parameter as input

```python
def repeat_calls(num_times):
    def decorator_function(func):
        @wraps(func)
        def wrapper(*args):
            for i in range(num_times):
                print(f"Executing {func.__name__} for iteration {i+1}")
                func(*args)
        return wrapper

    return decorator_function

@repeat_calls(num_times=3)
def greet(name):
    print("Hello {0}".format(name))
```

```python
greet("Shivam")
```

## Nesting Decorators
#### Below we nest stress_method with 2 decorators, @repeat_calls and @time_it
We call `stress_method(1.2)` once, it will be executed 3 times and once done time_it will tell total execution time of ~3.6 Sec

```python
@time_it
@repeat_calls(num_times=3)
def stress_method(seconds):
    time.sleep(seconds)

stress_method(1.2)
```
