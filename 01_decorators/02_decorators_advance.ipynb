### Decorator that tracks the count of Invoke
- This decorator increment the call count by 1, whenever the function is called.
- Useful to track how many time a function is called.
- Example: Track the number of times call to LLM is being made. 

```python
def call_counter(func):
    
    def wrapper(*args):
        wrapper.calls += 1
        return func(*args)

    wrapper.calls = 0
    return wrapper

@call_counter
def call_llm():
    print("Calling LLM !", call_llm.calls)
```

```python
for i in range(4):
    call_llm()
```

### Decorator that makes class as Singleton class
- @singleton decorator turns a class into a singleton by storing the first instance of the class as an attribute. 
- Later attempts at creating an instance simply return the stored instance cls.
- Exmple usecase, conncetion object to any Database, or Execution engine like Spark.

```python
def singleton(cls):  # cls is class reference
    def wrapper_singleton(*args, **kwargs):
        if not wrapper_singleton.instance:  # If isntance is not yet created
            wrapper_singleton.instance = cls(*args, **kwargs) # Create the object of class and store in variable
        return wrapper_singleton.instance # Always return same object on class object creation

    wrapper_singleton.instance = None # define variable that will hold the single object
    return wrapper_singleton

@singleton
class TestOne:
    pass
```

#### id of both objects will be same as they are referencing to same objectbecause of singletom behaviour


```python
first_one = TestOne()
another_one = TestOne() # will get the already created object, not creating new

print(id(first_one))
print(id(another_one))
```
