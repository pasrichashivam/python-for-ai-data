## Multi Threading
### Threadpool Executor - Thread based  parallelism

```python
import time
from concurrent.futures import ThreadPoolExecutor
```

```python
def fetch_data(url: str):
    print(f"Fetching data from {url}...")
    time.sleep(5)
    print(f"Data fetched from {url}.")
    return f"Data from {url}"

url_list = [
    'http://example.com/data1', 'http://example.com/data2','http://example.com/data3', 
    'http://example.com/data4', 'http://example.com/data5'
    ]
```

```python
results = []
with ThreadPoolExecutor(max_workers=len(url_list)) as executor:
    futures = list(executor.map(fetch_data, url_list))
    results.extend(futures)

for result in results:
    print(result)
```
